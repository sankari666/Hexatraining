```python
account_data = [
    (101, "John", "Chennai", 50000),
    (102, "Alice", "Bangalore", 75000),
    (103, "Bob", "Chennai", 30000),
    (104, "Emma", "Hyderabad", 45000),
    (105, "David", "Bangalore", 60000),
    (106, "Sophia", "Chennai", 25000)
]
columns = ["account_id", "customer_name", "city", "balance"]
```

```python
account_df = spark.createDataFrame(account_data, columns)
```

```python
display(account_df)
```

```python
city_balance_df = account_df.groupBy("city") \
                            .sum("balance") \
                            .withColumnRenamed("sum(balance)", "total_balance")
```

```python
display(city_balance_df)
print("Daily Banking Report Generated Successfully")
```
