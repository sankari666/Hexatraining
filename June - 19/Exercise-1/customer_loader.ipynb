```python
customer_data = [
    (1, "John", "Chennai"),
    (2, "Alice", "Bangalore"),
    (3, "Bob", "Chennai"),
    (4, "Emma", "Hyderabad"),
    (5, "David", "Bangalore"),
    (6, "Sophia", "Chennai")
]
```

```python
columns = ["customer_id", "customer_name", "city"]
customer_df = spark.createDataFrame(customer_data, columns)
```

```python
display(customer_df)
```

```python
customer_df.write.mode("overwrite").saveAsTable("customer_table")
print("customer_table created successfully")
```
