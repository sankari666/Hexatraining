```python
print("DEVICE ENERGY STATISTICS")
print("-" * 40)

for device in df['device_name'].unique():

    device_data = df[df['device_name'] == device]

    energy_array = device_data['energy_kwh'].to_numpy()

    total_energy = np.sum(energy_array)
    avg_energy = np.mean(energy_array)

    print(f"\nDevice: {device}")
    print("Total Energy:", round(total_energy, 2), "kWh")
    print("Average Energy:", round(avg_energy, 2), "kWh")
```

```python
room_summary = (
    df.groupby('room_name')['energy_kwh']
      .agg(['sum', 'mean', 'count'])
      .reset_index()
)

room_summary.columns = [
    'Room',
    'Total_Energy_kWh',
    'Average_Energy_kWh',
    'Reading_Count'
]

print("\nROOM SUMMARY")
print("-" * 40)
print(room_summary)
```

```python
device_summary = (
    df.groupby('device_name')['energy_kwh']
      .agg(['sum', 'mean', 'count'])
      .reset_index()
)

device_summary.columns = [
    'Device',
    'Total_Energy_kWh',
    'Average_Energy_kWh',
    'Reading_Count'
]

print("\nDEVICE SUMMARY")
print("-" * 40)
print(device_summary)
```

```python
df['date'] = df['timestamp'].dt.date

daily_summary = (
    df.groupby('date')['energy_kwh']
      .sum()
      .reset_index()
)

daily_summary.columns = [
    'Date',
    'Total_Energy_kWh'
]

print("\nDAILY ENERGY USAGE")
print("-" * 40)
print(daily_summary)
```
