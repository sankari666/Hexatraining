```python
import pandas as pd
import numpy as np
df = pd.read_csv("energy_usage.csv")
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
df['energy_kwh'] = pd.to_numeric(
    df['energy_kwh'],
    errors='coerce'
)
df = df.dropna(subset=['timestamp'])
mean_energy = df['energy_kwh'].mean()
df['energy_kwh'] = df['energy_kwh'].fillna(mean_energy)
```
