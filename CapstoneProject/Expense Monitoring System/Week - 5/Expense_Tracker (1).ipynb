```python
data = {
    "user_id": [101,101,101,102,102,103,104,104],
    "date": [
        "2026-06-01",
        "2026-06-10",
        "2026-07-02",
        "2026-06-05",
        "2026-07-04",
        "2026-06-03",
        "2026-07-08",
        "2026-07-10"
    ],
    "amount": [500,700,12000,1000,900,400,1500,20000]
}

df = pd.DataFrame(data)
df.to_csv("expenses.csv", index=False)

print("expenses.csv created successfully!\n")
```

```python
df = pd.read_csv("expenses.csv")

# Convert date column
df["date"] = pd.to_datetime(df["date"])

# Extract month
df["month"] = df["date"].dt.to_period("M")
```

```python
summary = df.groupby(
    ["user_id", "month"]
)["amount"].sum().reset_index()

summary.rename(columns={
    "user_id":"User_ID",
    "month":"Month",
    "amount":"Monthly_Spend"
}, inplace=True)
```

```python
BUDGET = 20000

summary["Savings"] = BUDGET - summary["Monthly_Spend"]
```

```python
def get_alert(spend):
    if spend > 18000:
        return "High Spending"
    elif spend > 10000:
        return "Medium Spending"
    else:
        return "Normal"

summary["Alert"] = summary["Monthly_Spend"].apply(get_alert)
```

```python
summary.to_csv("monthly_analysis.csv", index=False)
```

```python
print("Monthly Expense Report")
print(summary)

print("\nSavings Alerts")

for _, row in summary.iterrows():
    if row["Monthly_Spend"] > 18000:
        print(f"ALERT: User {row['User_ID']} exceeded the budget in {row['Month']} (Spent ₹{row['Monthly_Spend']})")

print("\nAnalysis Completed Successfully!")
```
