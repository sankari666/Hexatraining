```python
appointments_df = spark.read \
.option("header","true") \
.option("inferSchema","true") \
.csv("/Volumes/hexa_databrick_7405612841012936/default/hexa-volume/appointments.csv")

appointments_df.show()

appointments_df.printSchema()
```

```python
(
    appointments_df.write
    .mode("overwrite")
    .format("delta")
    .save("/Volumes/hexa_databrick_7405612841012936/default/delta-appointment")
)
```

```python
spark.read.format("delta") \
.load("/Volumes/hexa_databrick_7405612841012936/default/delta-appointment") \
.show()
```
