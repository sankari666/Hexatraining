```python
from pyspark.sql.functions import col

preferences_df = spark.read \
.option("multiline","true") \
.json("/Volumes/hexa_databrick_7405612841012936/default/hexa-volume/patient_preferences.json")

preferences_df.show()
preferences_df.printSchema()
```

```python
preferences_flat_df = preferences_df.select(
    "patient_id",
    "preferred_hospital",
    col("contact.phone").alias("phone"),
    col("contact.email").alias("email")
)

preferences_flat_df.show()
```

```python
(
    preferences_flat_df.write
    .mode("overwrite")
    .format("delta")
    .save("/Volumes/hexa_databrick_7405612841012936/default/delta-json")
)
```

```python
spark.read.format("delta") \
.load("/Volumes/hexa_databrick_7405612841012936/default/delta-json") \
.show()
```
