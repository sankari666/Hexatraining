```python
#76-90 Window functions
from pyspark.sql.window import Window
from pyspark.sql.functions import *
```

```python
window_spec = Window.orderBy(
    col("total_amount").desc()
)

df.withColumn(
    "row_number",
    row_number().over(window_spec)
).show()

df.withColumn(
    "rank",
    rank().over(window_spec)
).show()

df.withColumn(
    "rank",
    rank().over(window_spec)
).show()

df.withColumn(
    "dense_rank",
    dense_rank().over(window_spec)
).show()

df.withColumn(
    "rank",
    rank().over(window_spec)
).filter(
    col("rank") == 1
).show()



```

```python
city_window = Window.partitionBy(
    "city"
).orderBy(
    col("total_amount").desc()
)

df.withColumn(
    "rank",
    rank().over(city_window)
).filter(
    col("rank") == 1
).show()

category_window = Window.partitionBy(
    "category"
).orderBy(
    col("total_amount").desc()
)

df.withColumn(
    "rank",
    rank().over(category_window)
).filter(
    col("rank") == 1
).show()


running_window = Window.orderBy(
    "order_id"
)

df.withColumn(
    "running_total",
    sum("total_amount").over(running_window)
).show()


df.withColumn(
    "previous_revenue",
    lag("total_amount").over(window_spec)
).show()

compare_df = df.withColumn(
    "previous_revenue",
    lag("total_amount").over(window_spec)
)

compare_df.select(
    "order_id",
    "total_amount",
    "previous_revenue"
).show()



compare_df = df.withColumn(
    "previous_revenue",
    lag("total_amount").over(window_spec)
)

compare_df.withColumn(
    "revenue_difference",
    col("total_amount") - col("previous_revenue")
).show()

df.withColumn(
    "rank",
    rank().over(window_spec)
).select(
    "order_id",
    "product",
    "city",
    "category",
    "total_amount",
    "rank"
).show()
```

```python
#92
df.select(
    "product",
    "quantity",
    "unit_price",
    "total_amount",
    "tax",
    "discount",
    "net_amount"
).show()

#93
completed_df = df.filter(
    col("order_status") == "Completed"
)
completed_df.show()

#94
completed_df.agg(
    sum("total_amount")
    .alias("total_revenue")
).show()

#95
city_revenue_df = completed_df.groupBy(
    "city"
).agg(
    sum("total_amount")
    .alias("city_revenue")
)
city_revenue_df.show()

#96
category_revenue_df = completed_df.groupBy(
    "category"
).agg(
    sum("total_amount")
    .alias("category_revenue")
)
category_revenue_df.show()

#97
completed_df.write.mode(
    "overwrite"
).saveAsTable(
    "databrick_hexa.default.completed_sales"
)

#98
parquet_df = spark.read.table(
    "databrick_hexa.default.completed_sales"
)
parquet_df.show()

#99
parquet_df.createOrReplaceTempView(
    "sales_parquet"
)



```
