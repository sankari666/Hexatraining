```python
sales_rdd = spark.sparkContext.parallelize(numbers)
```

```python
total_sum = sales_rdd.reduce(lambda a, b: a + b)
print(total_sum)
```

```python
max_value = sales_rdd.reduce(lambda a, b: a if a > b else b)
print(max_value)
```

```python
min_value = sales_rdd.reduce(lambda a, b: a if a < b else b)
print(min_value)
```

```python
total = sales_rdd.reduce(lambda a, b: a + b)
count = sales_rdd.count()
average = total / count
print(average)
```

```python
count_gt_1000 = sales_rdd.filter(lambda x: x > 1000).count()
print(count_gt_1000)
```

```python
doubled_rdd = sales_rdd.map(lambda x: x * 2)
print(doubled_rdd.collect())
```

```python
plus_100_rdd = sales_rdd.map(lambda x: x + 100)
print(plus_100_rdd.collect())
```

```python
sorted_asc_rdd = sales_rdd.sortBy(lambda x: x, ascending=True)
print(sorted_asc_rdd.collect())
```

```python
courses = [
"Python SQL Spark",
"Azure Databricks Spark",
"Python Pandas NumPy",
"SQL Spark DeltaLake",
"Azure Python SQL"
]
```

```python
course_rdd = spark.sparkContext.parallelize(courses)
```

```python
words_rdd =course_rdd.flatMap(lambda line: line.split(" "))
print(words_rdd.collect())
```

```python
total_words = words_rdd.count()
print(total_words)
```

```python
distinct_words_rdd = words_rdd.distinct()
print(distinct_words_rdd.collect())
```

```python
sorted_words_rdd = words_rdd.distinct().sortBy(lambda word: word.lower())
print(sorted_words_rdd.collect())
```

```python
word_frequencies = words_rdd.countByValue()
print(dict(word_frequencies))
```

```python
python_words = words_rdd.distinct().filter(lambda word: "Python" in word)
print(python_words.collect())
```

```python
spark_words = words_rdd.distinct().filter(lambda word: "Spark" in word)
print(spark_words.collect())
```

```python
uppercase_words_rdd = words_rdd.map(lambda word: word.upper())
print(uppercase_words_rdd.distinct().collect())
```
