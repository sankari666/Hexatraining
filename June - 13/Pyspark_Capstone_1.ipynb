  #58

```python
daily_revenue = retail_sales_df.groupBy(
    "sale_date"
).agg(
    sum("sale_amount").alias("daily_revenue")
)
```

```python
running_window = Window.orderBy("sale_date")

daily_revenue = daily_revenue.withColumn(
    "running_total_revenue",
    sum("daily_revenue").over(
        running_window.rowsBetween(
            Window.unboundedPreceding,
            Window.currentRow
        )
    )
)
```

  #59

```python
daily_revenue = daily_revenue.withColumn(
    "previous_day_revenue",
    lag("daily_revenue").over(running_window)
)
daily_revenue = daily_revenue.withColumn(
    "difference",
    col("daily_revenue") - col("previous_day_revenue")
)
daily_revenue.show()
```
