```python
from pyspark.sql.functions import col, when
df_report = df_fin.withColumn(
    "discount",
    col("total_amount") * 0.10
).withColumn(
    "final_amount",
    col("total_amount") - col("discount")
)
df_report = df_report.withColumn(
    "order_value_category",
    when(col("total_amount") >= 50000, "High")
    .when(col("total_amount") >= 20000, "Medium")
    .otherwise("Low")
)
df_report.show()
```

```python
df_report.groupBy("order_value_category").count().show()
```

```python
df_report.filter(col("order_status") == "delivered") \
         .select("order_id", "product", "order_status", "final_amount") \
         .show()
```

```python
from pyspark.sql.functions import sum
df_report.filter(col("order_status") == "delivered") \
         .groupBy("city") \
         .agg(sum("final_amount").alias("delivered_revenue")) \
         .show()
```

```python
final_report_df = df_report.select(
    "order_id",
    "customer_name",
    "city",
    "category",
    "product",
    "quantity",
    "price",
    "total_amount",
    "discount",
    "final_amount",
    "order_status"
)
final_report_df.show()
```
